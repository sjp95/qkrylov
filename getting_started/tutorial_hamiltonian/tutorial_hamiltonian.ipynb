# Building the Hamiltonian

Welcome to the quickstart. In this tutorial, we will construct a 1D Heisenberg Spin-1/2 chain. 


## Step 1: The Hilbert Space

First, we define our Hilbert space. We use `SpinHalfBasis`.

```python
import qkrylov as qk

b = qk.SpinHalfBasis(10, sz=0)
print(f"Basis dimension: {b.dimension}")
```

## Step 2: Operator Sum

Next, we use `OpSum` to symbolically build our interactions.

```python
ops = qk.OpSum()
J = 1.0
for i in range(9):
    ops += (J, "Sz", i, "Sz", i + 1)
    ops += (J*0.5, "Sp", i, "Sm", i + 1)
    ops += (J*0.5, "Sm", i, "Sp", i + 1)
```

## Step 3: Solve

Finally, we construct the matrix-free Hamiltonian and solve it.

```python
H = qk.MatrixFreeHamiltonian(b, ops, device="cpu")
result = qk.lanczos_ground_state(H)
print(f"Ground State Energy: {result.energy}")
```
